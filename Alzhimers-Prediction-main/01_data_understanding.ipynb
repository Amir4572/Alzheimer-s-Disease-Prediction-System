The Alzheimer's disease dataset contains clinical and demographic information about patients. This data will be used to predict whether a patient has Alzheimer's disease based on various health indicators and lifestyle factors.

```python
df = pd.read_csv('alzheimers_disease_data.csv')
print("Dataset loaded successfully!")
print(f"Number of records: {df.shape[0]}")
print(f"Number of features: {df.shape[1]}")
```

## Checking for Missing Values

Missing values can significantly impact the performance of machine learning models. It is important to identify any missing data early so we can decide on appropriate handling strategies during preprocessing.

```python
missing_values = df.isnull().sum()
print("Missing Values per Column:")
print("="*60)
print(missing_values[missing_values > 0] if missing_values.sum() > 0 else "No missing values found in the dataset!")
print(f"\nTotal missing values: {missing_values.sum()}")
```

## Target Variable Analysis

The target variable in this dataset is 'Diagnosis', which indicates whether a patient has Alzheimer's disease (1) or not (0). Understanding the distribution of this variable is essential as it affects how we train and evaluate our models.

```python
print("Target Variable Distribution:")
print("="*60)
diagnosis_counts = df['Diagnosis'].value_counts()
diagnosis_percentages = df['Diagnosis'].value_counts(normalize=True) * 100

print(f"No Alzheimer's (0): {diagnosis_counts[0]} patients ({diagnosis_percentages[0]:.2f}%)")
print(f"Alzheimer's (1):    {diagnosis_counts[1]} patients ({diagnosis_percentages[1]:.2f}%)")
```

```python
plt.figure(figsize=(8, 6))
colors = ['#3498db', '#e74c3c']
plt.pie(diagnosis_counts, labels=['No Alzheimer\'s', 'Alzheimer\'s'], autopct='%1.1f%%', 
        colors=colors, startangle=90, explode=(0.02, 0.02))
plt.title('Distribution of Diagnosis', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Observations on Class Balance

The dataset shows an imbalanced distribution with more patients without Alzheimer's disease compared to those with the condition. This class imbalance is important to address during preprocessing, as it can cause models to be biased toward the majority class. We will use techniques like SMOTE (Synthetic Minority Over-sampling Technique) in the preprocessing notebook to handle this issue.

## Feature Categories

The features in this dataset can be broadly categorized into several groups based on their nature and the type of information they capture.

```python
demographic_features = ['Age', 'Gender', 'Ethnicity', 'EducationLevel']
lifestyle_features = ['BMI', 'Smoking', 'AlcoholConsumption', 'PhysicalActivity', 'DietQuality', 'SleepQuality']
medical_history = ['FamilyHistoryAlzheimers', 'CardiovascularDisease', 'Diabetes', 'Depression', 'HeadInjury', 'Hypertension']
clinical_measurements = ['SystolicBP', 'DiastolicBP', 'CholesterolTotal', 'CholesterolLDL', 'CholesterolHDL', 'CholesterolTriglycerides']
cognitive_assessments = ['MMSE', 'FunctionalAssessment', 'MemoryComplaints', 'BehavioralProblems', 'ADL']
symptoms = ['Confusion', 'Disorientation', 'PersonalityChanges', 'DifficultyCompletingTasks', 'Forgetfulness']

print("Feature Categories:")
print("="*60)
print(f"\nDemographic Features ({len(demographic_features)}):")
print(f"  {', '.join(demographic_features)}")
print(f"\nLifestyle Features ({len(lifestyle_features)}):")
print(f"  {', '.join(lifestyle_features)}")
print(f"\nMedical History ({len(medical_history)}):")
print(f"  {', '.join(medical_history)}")
print(f"\nClinical Measurements ({len(clinical_measurements)}):")
print(f"  {', '.join(clinical_measurements)}")
print(f"\nCognitive Assessments ({len(cognitive_assessments)}):")
print(f"  {', '.join(cognitive_assessments)}")
print(f"\nSymptoms ({len(symptoms)}):")
print(f"  {', '.join(symptoms)}")
```

## Age Distribution Analysis

Age is one of the most significant risk factors for Alzheimer's disease. Let us examine how age is distributed in our dataset and how it relates to the diagnosis.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['Age'], bins=20, color='steelblue', edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Age', fontsize=12)
axes[0].set_ylabel('Frequency', fontsize=12)
axes[0].set_title('Age Distribution', fontsize=14, fontweight='bold')
axes[0].axvline(df['Age'].mean(), color='red', linestyle='--', label=f'Mean: {df["Age"].mean():.1f}')
axes[0].legend()

colors = ['#3498db', '#e74c3c']
for i, diagnosis in enumerate([0, 1]):
    subset = df[df['Diagnosis'] == diagnosis]['Age']
    axes[1].hist(subset, bins=20, alpha=0.6, label=f'Diagnosis {diagnosis}', color=colors[i], edgecolor='black')
axes[1].set_xlabel('Age', fontsize=12)
axes[1].set_ylabel('Frequency', fontsize=12)
axes[1].set_title('Age Distribution by Diagnosis', fontsize=14, fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Correlation Analysis

Understanding the correlations between features helps us identify which variables might be most predictive of Alzheimer's disease and detect any highly correlated features that might cause multicollinearity issues.

```python
numeric_df = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(18, 14))
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', center=0, 
            square=True, linewidths=0.5)
plt.title('Correlation Heatmap of All Features', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Features Most Correlated with Diagnosis

Let us identify which features have the strongest correlations with the target variable. This gives us an initial indication of which features might be most useful for prediction.

```python
diagnosis_correlations = correlation_matrix['Diagnosis'].drop('Diagnosis').sort_values(key=abs, ascending=False)

print("Features Correlated with Diagnosis (sorted by absolute value):")
print("="*60)
for feature, corr in diagnosis_correlations.items():
    print(f"{feature:30} : {corr:+.4f}")
```

```python
plt.figure(figsize=(12, 8))
colors = ['#e74c3c' if x < 0 else '#3498db' for x in diagnosis_correlations.values]
plt.barh(range(len(diagnosis_correlations)), diagnosis_correlations.values, color=colors)
plt.yticks(range(len(diagnosis_correlations)), diagnosis_correlations.index)
plt.xlabel('Correlation Coefficient', fontsize=12)
plt.title('Feature Correlations with Diagnosis', fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linewidth=0.5)
plt.tight_layout()
plt.show()
```

## Key Cognitive and Functional Features

Cognitive assessments like MMSE (Mini-Mental State Examination), Functional Assessment, and ADL (Activities of Daily Living) scores are particularly important in Alzheimer's diagnosis. Let us examine these features more closely.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

cognitive_features = ['MMSE', 'FunctionalAssessment', 'ADL']
titles = ['MMSE Score', 'Functional Assessment', 'ADL Score']

for idx, (feature, title) in enumerate(zip(cognitive_features, titles)):
    for diagnosis in [0, 1]:
        subset = df[df['Diagnosis'] == diagnosis][feature]
        label = 'No Alzheimer\'s' if diagnosis == 0 else 'Alzheimer\'s'
        axes[idx].hist(subset, bins=15, alpha=0.6, label=label, edgecolor='black')
    axes[idx].set_xlabel(title, fontsize=12)
    axes[idx].set_ylabel('Frequency', fontsize=12)
    axes[idx].set_title(f'{title} by Diagnosis', fontsize=12, fontweight='bold')
    axes[idx].legend()

plt.tight_layout()
plt.show()
```

## Lifestyle Factors Distribution

Lifestyle factors such as BMI, physical activity, diet quality, and sleep quality may also play a role in Alzheimer's disease risk. Let us visualize how these factors are distributed across our dataset.

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, feature in enumerate(lifestyle_features):
    axes[idx].hist(df[feature], bins=20, color='teal', edgecolor='black', alpha=0.7)
    axes[idx].set_xlabel(feature, fontsize=10)
    axes[idx].set_ylabel('Frequency', fontsize=10)
    axes[idx].set_title(f'Distribution of {feature}', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

## Binary Features Analysis

Several features in the dataset are binary (0 or 1), representing the presence or absence of certain conditions or symptoms. Let us examine their distribution and relationship with the diagnosis.

```python
binary_features = medical_history + symptoms

fig, axes = plt.subplots(3, 4, figsize=(16, 12))
axes = axes.flatten()

for idx, feature in enumerate(binary_features):
    if idx < len(axes):
        cross_tab = pd.crosstab(df[feature], df['Diagnosis'], normalize='index') * 100
        cross_tab.plot(kind='bar', ax=axes[idx], color=['#3498db', '#e74c3c'], edgecolor='black')
        axes[idx].set_title(feature, fontsize=10, fontweight='bold')
        axes[idx].set_xlabel('')
        axes[idx].set_ylabel('Percentage')
        axes[idx].legend(['No AD', 'AD'], loc='upper right')
        axes[idx].tick_params(axis='x', rotation=0)

for idx in range(len(binary_features), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```
