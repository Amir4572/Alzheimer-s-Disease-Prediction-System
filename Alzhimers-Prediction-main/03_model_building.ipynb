# Notebook 3: Model Building

This notebook focuses on training machine learning models to predict Alzheimer's disease. We will train six different algorithms, each with its own strengths and approach to classification. Using multiple models allows us to compare their performance and select the best approach for our prediction task.

The models we train here will be evaluated in detail in the next notebook.

## Loading Required Libraries

We import the machine learning models from scikit-learn along with utilities for saving trained models. Each model represents a different approach to classification, from simple linear models to more complex ensemble methods.

```python
import numpy as np
import pandas as pd
import pickle
import warnings

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

warnings.filterwarnings('ignore')
```

## Loading Preprocessed Data

We load the preprocessed data that was saved in the previous notebook. This includes the balanced training data (after SMOTE) and the scaled test data. Using preprocessed data ensures consistency and saves us from repeating the preprocessing steps.

```python
X_train = np.load('X_train_balanced.npy')
y_train = np.load('y_train_balanced.npy')
X_test = np.load('X_test_scaled.npy')
y_test = np.load('y_test.npy')

feature_names = pd.read_csv('feature_names.csv')['feature_name'].tolist()

print("Preprocessed data loaded successfully!")
print("="*50)
print(f"Training samples: {X_train.shape[0]}")
print(f"Testing samples:  {X_test.shape[0]}")
print(f"Number of features: {X_train.shape[1]}")
```

## Model 1: Logistic Regression

Logistic Regression is a simple yet powerful algorithm for binary classification. Despite its name, it is a classification algorithm that estimates the probability of a sample belonging to a particular class. It works well when there is a linear relationship between features and the log-odds of the outcome.

We use Logistic Regression as our baseline model because of its simplicity and interpretability. The model's coefficients can tell us which features have the most influence on the prediction.

```python
print("Training Logistic Regression...")
logistic_model = LogisticRegression(random_state=42, max_iter=1000)
logistic_model.fit(X_train, y_train)
print("Logistic Regression training complete!")
```

## Model 2: Decision Tree

Decision Tree is a rule-based classification algorithm that creates a tree structure where each internal node represents a decision based on a feature value. It is highly interpretable because we can visualize the decision rules as a flowchart.

Decision Trees are excellent for understanding the decision-making process and work well with both numerical and categorical data. However, they can be prone to overfitting if not properly constrained.

```python
print("Training Decision Tree...")
decision_tree_model = DecisionTreeClassifier(random_state=42, max_depth=10)
decision_tree_model.fit(X_train, y_train)
print("Decision Tree training complete!")
```

## Model 3: Random Forest

Random Forest is an ensemble method that combines multiple decision trees to create a stronger, more robust classifier. Each tree in the forest is trained on a random subset of the data and features, and the final prediction is made by majority voting.

This approach reduces overfitting and generally provides better performance than a single decision tree. Random Forest also provides feature importance scores, which help us understand which features contribute most to the predictions.

```python
print("Training Random Forest...")
random_forest_model = RandomForestClassifier(random_state=42, n_estimators=100)
random_forest_model.fit(X_train, y_train)
print("Random Forest training complete!")
```

## Model 4: Support Vector Machine (SVM)

Support Vector Machine is a powerful algorithm that finds the optimal hyperplane to separate different classes. SVM is particularly effective in high-dimensional spaces and works well when the number of features is large relative to the number of samples.

We use the RBF (Radial Basis Function) kernel, which allows SVM to capture non-linear relationships in the data. The probability parameter is enabled to allow SVM to output probability estimates.

```python
print("Training Support Vector Machine...")
svm_model = SVC(random_state=42, kernel='rbf', probability=True)
svm_model.fit(X_train, y_train)
print("Support Vector Machine training complete!")
```

## Model 5: k-Nearest Neighbors (KNN)

k-Nearest Neighbors is an instance-based learning algorithm that classifies new samples based on the majority class of their k nearest neighbors in the feature space. It is a simple but effective algorithm that makes no assumptions about the underlying data distribution.

KNN stores all training data and makes predictions by measuring distances to find similar samples. We use k=5 neighbors, which is a common default that balances between capturing local patterns and avoiding noise.

```python
print("Training k-Nearest Neighbors...")
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
print("k-Nearest Neighbors training complete!")
```

## Model 6: Naive Bayes

Naive Bayes is a probabilistic classifier based on Bayes' theorem with the assumption that features are independent of each other (hence "naive"). Despite this simplifying assumption, Naive Bayes often performs surprisingly well in practice.

Gaussian Naive Bayes assumes that the features follow a normal distribution. It is fast to train and works well with continuous data, making it suitable for our standardized features.

```python
print("Training Naive Bayes...")
naive_bayes_model = GaussianNB()
naive_bayes_model.fit(X_train, y_train)
print("Naive Bayes training complete!")
```

## Saving Trained Models

We save all trained models to files using Python's pickle module. This allows us to load the models in subsequent notebooks without retraining them, saving time and ensuring reproducibility.

```python
models = {
    'Logistic Regression': logistic_model,
    'Decision Tree': decision_tree_model,
    'Random Forest': random_forest_model,
    'Support Vector Machine': svm_model,
    'K-Nearest Neighbors': knn_model,
    'Naive Bayes': naive_bayes_model
}

with open('trained_models.pkl', 'wb') as f:
    pickle.dump(models, f)

print("All models saved successfully!")
print("="*50)
print("File created: trained_models.pkl")
```

## Quick Performance Check

Before moving to detailed evaluation, let us check the training and testing accuracy of each model to get a preliminary understanding of their performance.

```python
print("Model Performance Overview:")
print("="*60)
print(f"{'Model':<25} {'Train Acc':<12} {'Test Acc':<12}")
print("-"*60)

for name, model in models.items():
    train_acc = model.score(X_train, y_train) * 100
    test_acc = model.score(X_test, y_test) * 100
    print(f"{name:<25} {train_acc:>8.2f}%    {test_acc:>8.2f}%")
```

## Feature Importance from Random Forest

Random Forest provides feature importance scores that indicate how much each feature contributes to the model's predictions. Features with higher importance scores are more influential in determining whether a patient has Alzheimer's disease.

```python
feature_importance = pd.DataFrame({
    'Feature': feature_names,
    'Importance': random_forest_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("Top 10 Most Important Features (Random Forest):")
print("="*50)
for idx, row in feature_importance.head(10).iterrows():
    print(f"{row['Feature']:<30} {row['Importance']:.4f}")
```

We have successfully trained six different machine learning models, each with its unique approach to classification:
